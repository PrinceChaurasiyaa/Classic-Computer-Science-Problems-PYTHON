# Fibonacci sequence

## The Fibonacci sequence is a sequence of numbers such that any number, except for the first and second, is the sum of the previous two:
## 0, 1, 1, 2, 3, 5, 8, 13, 21...

## fib(n) = fib(n - 1) + fib(n - 2)

### The preceding formula for computing a number in the Fibonacci sequence is a form of pseudocode that can be trivially translated into a recursive Python function. (A recursive function is a function that calls itself.)

```python
# def fib1(n: int) -> int:
#     return fib1(n-1) + fib(n-2)



# if __name__ == "__main__":
#     print(fib1(3))


"""
If we try to run fib1.py, we generate an error:
RecursionError: maximum recursion depth exceeded

due to infinite recursion
"""
```

```python
def fib2(n: int) -> int:
    if n < 2:      # base case
        return n
    
    return fib2(n-2) + fib2(n-1)


if __name__ == "__main__":
    print(fib2(5))
```

# Memoization
### Memoization is an optimization technique that stores the results of expensive function calls in a cache and returns the cached result when the same inputs occur again
#### Donald Michie, a famous British computer scientist, coined the term memoization. Memo functions: a language feature with “rote-learning” properties

```python
from typing import Dict
memo: Dict[int, int] = {0: 0, 1:1}

def fib3(n: int) -> int:
    if n not in memo:
        memo[n] = fib3(n-2) + fib3(n-1)
    
    return memo[n]


if __name__ == "__main__":
    print(fib3(15))
```

### What is a decorator? 
#### A decorator is a function that modifies or extends the behavior of another function without changing its original code.

# Automatic memoization
### Python has a built-in decorator for memoizing any function automagically. decorator @functools.lru_cache()
#### "@functools.lru_cache() is a decorator in Python that caches the results of function calls. It avoids repeated calculations when the same arguments are passed to a function. It is particularly useful in recursive problems and dynamic programming, such as Fibonacci. LRU stands for Least Recently Used, which determines which cached entries are removed when the cache reaches its maximum size." it remembers the results of previous function calls.

```python
from functools import lru_cache

@lru_cache(maxsize=None)
def fib4(n: int) -> int:
    if n < 2:
        return n
    return fib4(n-2) + fib4(n-1)


if __name__ == "__main__":
    print(fib4(5))
    print(fib4(15))
```
