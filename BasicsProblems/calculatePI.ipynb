# Calculating pi

The mathematically significant number pi (π or 3.14159…) can be derived using
many formulas. 

One of the simplest is the Leibniz formula. It posits that the conver
gence of the following infinite series is equal to pi:
π = 4/1 - 4/3 + 4/5 - 4/7 + 4/9 - 4/11...

```python
def calculatePI(nTerms: int) -> float:
    numerator: float = 4.0
    denominator: float = 1.0
    operation: float = 1.0
    pi: float = 0.0

    for _ in range(nTerms):
        pi += operation * (numerator / denominator)
        denominator += 2.0
        operation *= -1.0
    
    return pi
```

```python
if __name__ == "__main__":
    print(calculatePI(1000000000))
```
